```python
# Importing additional libraries for multivariate analysis
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from statsmodels.stats.outliers_influence import variance_inflation_factor
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
df.fillna(df.mean(), inplace=True)
```

# Feature Scaling

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df.select_dtypes(include=np.number))
```

# Correlation Matrix for detecting multicollinearity

```python
corr_matrix = df.corr()
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()
```

# Detecting multicollinearity using VIF (Variance Inflation Factor)

```python
numeric_cols = df.select_dtypes(include=np.number).columns


scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[numeric_cols])

vif_data = pd.DataFrame()
vif_data["Feature"] = numeric_cols 
vif_data["VIF"] = [variance_inflation_factor(scaled_data, i) for i in range(scaled_data.shape[1])]

print(vif_data)

features_to_drop = vif_data[vif_data['VIF'] > 10]['Feature']
print(f"Features with VIF > 10: {features_to_drop.tolist()}")

if not features_to_drop.empty:
    df.drop(columns=features_to_drop, inplace=True)
    print(f"Features dropped: {features_to_drop.tolist()}")
else:
    print("No features with VIF > 10 found.")
```

# t-SNE for Dimensionality Reduction

```python
tsne = TSNE(n_components=2, perplexity=30, n_iter=300)
tsne_result = tsne.fit_transform(scaled_data)
df['t-SNE1'] = tsne_result[:, 0]
df['t-SNE2'] = tsne_result[:, 1]
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(df['t-SNE1'], df['t-SNE2'], c='green', marker='o')
plt.title('t-SNE - 2D Projection')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')
plt.grid(True)
plt.show()
```

# Principal Component Analysis (PCA) for Dimensionality Reduction

```python
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)
df['PCA1'] = pca_result[:, 0]
df['PCA2'] = pca_result[:, 1]
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(df['PCA1'], df['PCA2'], c='blue', marker='o')
plt.title('PCA - 2D Projection')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```

# Clustering with KMeans

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(scaled_data)
```

```python
silhouette_avg = silhouette_score(scaled_data, df['Cluster'])
print(f'Silhouette Score for the Clusters: {silhouette_avg}')
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster', data=df, palette='viridis')
plt.title('Clusters Visualized in PCA space')
plt.show()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df_numeric = df.select_dtypes(include=['float64', 'int64'])

scaler = StandardScaler()
df_scaled = scaler.fit_transform(df_numeric)

pca = PCA(n_components=2)  # Reduce to 2 components for visualization
pca_components = pca.fit_transform(df_scaled)

plt.figure(figsize=(10, 6))
plt.scatter(pca_components[:, 0], pca_components[:, 1])
plt.title('PCA: First two components')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()

print(f'Explained variance by component: {pca.explained_variance_ratio_}')
```
